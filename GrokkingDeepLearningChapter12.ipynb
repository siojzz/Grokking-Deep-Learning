# Chapter 12 - Neural Networks That Write Like Shakespeare

This chapter introduces Recurrent Neural Networks (RNNs).

RNNs are designed for sequence data and can use information from previous steps when processing the current input.

```python
import numpy as np

np.random.seed(1)

input_size = 3
hidden_size = 4

Wx = np.random.randn(
    input_size,
    hidden_size
) * 0.1

Wh = np.random.randn(
    hidden_size,
    hidden_size
) * 0.1


hidden = np.zeros(
    (1, hidden_size)
)

sequence = [
    np.array([[1, 0, 0]]),
    np.array([[0, 1, 0]]),
    np.array([[0, 0, 1]])
]


for step, x in enumerate(
    sequence
):

    hidden = np.tanh(
        x.dot(Wx)
        + hidden.dot(Wh)
    )

    print(
        "Step",
        step,
        hidden
    )
```

**Interpretation:**

The hidden state stores information from previous sequence steps.

The current hidden state depends on both the current input and the previous hidden state.

This gives RNNs a form of memory.

## Chapter Summary

In this chapter, I learned how recurrent neural networks process sequence data.

This allows them to model information that depends on previous elements in a sequence.
