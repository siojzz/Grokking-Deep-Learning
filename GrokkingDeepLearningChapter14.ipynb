## 1. Vanishing and Exploding Gradients

During backpropagation through long sequences, gradients can repeatedly become smaller or larger.

Very small gradients can prevent learning. This is called the vanishing gradient problem.

Very large gradients can make training unstable. This is called the exploding gradient problem.

```python
gradient = 1.0

for step in range(10):

    gradient *= 1.5

    print(
        "Step:",
        step,
        "Gradient:",
        gradient
    )
```

## 2. LSTM

An LSTM uses gates to control information stored in its cell state.

Important gates include:

- Forget gate
- Input gate
- Output gate

These gates help the network retain useful information for longer periods.

```python
import numpy as np

def sigmoid(x):
    return 1 / (
        1 + np.exp(-x)
    )


x = np.array([
    [0.5, -0.2]
])

previous_hidden = np.array([
    [0.1, 0.3]
])

combined = np.concatenate(
    [x, previous_hidden],
    axis=1
)

np.random.seed(1)

W = np.random.randn(
    combined.shape[1],
    2
)

forget_gate = sigmoid(
    combined.dot(W)
)

print(
    "Forget gate:",
    forget_gate
)
```

**Interpretation:**

The gate produces values between 0 and 1.

These values control how much information should be kept or forgotten.

LSTMs use several gates together to manage long-term information.
