# Chapter 15 - Deep Learning on Unseen Data: Federated Learning

This chapter introduces Federated Learning.

Federated Learning allows multiple devices or data sources to train models without sending all raw data to one central location.

## 1. Centralized Learning

In traditional centralized learning, training data is collected in one location.

The model is trained directly using all of this data.

This approach may create privacy and data ownership concerns.

## 2. Federated Learning

In federated learning, each client keeps its own data.

The clients train local versions of a model and send model updates instead of raw training data.

A central server combines the updates to create a new global model.

```python
import numpy as np

client_1_weights = np.array([
    0.5,
    0.2,
    0.1
])

client_2_weights = np.array([
    0.4,
    0.3,
    0.2
])

client_3_weights = np.array([
    0.6,
    0.1,
    0.15
])


global_weights = np.mean(
    [
        client_1_weights,
        client_2_weights,
        client_3_weights
    ],
    axis=0
)

print(
    "Global weights:",
    global_weights
)
```

**Interpretation:**

Each client trains a local model using its own data.

The local model parameters are then combined to update the global model.

The raw data itself does not need to be collected in one central dataset.

## 3. Privacy

Federated learning can reduce the need to centralize sensitive data.

However, federated learning alone does not automatically guarantee complete privacy.

Additional privacy and security techniques may still be necessary.

## Chapter Summary

In this chapter, I learned the basic concept of federated learning.

Instead of sending raw data to a central server, clients can train models locally and share model updates.

The updates can then be aggregated into a global model.

Federated learning is useful when data is distributed across many devices or organizations.
