# Chapter 7 - How to Picture Neural Networks

This chapter focuses on developing an intuitive way to visualize neural networks.

Visualizing the network makes it easier to understand how information and error signals move through different layers.

## 1. Correlation Summarization

A neural network can be viewed as a system that learns useful relationships or correlations between input and output variables.

Weights represent the strength and direction of these relationships.

## 2. Visualizing Layers

A simple neural network can be represented as:

Input Layer → Hidden Layer → Output Layer

Each connection has a weight.

Forward propagation moves information from input to output, while backpropagation sends error information in the opposite direction.

```python
network = {
    "Input Layer": 3,
    "Hidden Layer": 4,
    "Output Layer": 1
}

for layer, neurons in network.items():
    print(layer, ":", neurons, "neurons")
```
