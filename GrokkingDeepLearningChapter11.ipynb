# Chapter 11 - Neural Networks That Understand Language

This chapter introduces word embeddings.

Word embeddings represent words as numerical vectors that can be learned by neural networks.

```python
import numpy as np

words = [
    "king",
    "queen",
    "man",
    "woman"
]

word2index = {
    word: i
    for i, word
    in enumerate(words)
}

np.random.seed(1)

embedding_size = 3

embeddings = (
    np.random.randn(
        len(words),
        embedding_size
    )
)

for word in words:
    print(
        word,
        embeddings[
            word2index[word]
        ]
    )
```

## Cosine Similarity

Words with similar vector representations can have high cosine similarity.

This provides a way to compare relationships between learned word embeddings.

```python
def cosine_similarity(a, b):

    return np.dot(a, b) / (
        np.linalg.norm(a)
        * np.linalg.norm(b)
    )


king = embeddings[
    word2index["king"]
]

queen = embeddings[
    word2index["queen"]
]

print(
    "Similarity:",
    cosine_similarity(
        king,
        queen
    )
)
```
