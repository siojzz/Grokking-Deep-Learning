# Chapter 8 - Learning Signal and Ignoring Noise

This chapter introduces overfitting, regularization, dropout, and batching.

The goal is to help neural networks learn useful patterns instead of memorizing noise from training data.

## 1. Overfitting

Overfitting happens when a model performs very well on training data but performs poorly on new data.

The network has learned details or noise that do not generalize.

```python
import numpy as np

np.random.seed(1)

layer = np.array([
    [0.5, 0.3, 0.2, 0.8]
])

dropout_mask = np.random.randint(
    2,
    size=layer.shape
)

layer_dropout = (
    layer
    * dropout_mask
    * 2
)

print("Original:")
print(layer)

print("\nDropout mask:")
print(dropout_mask)

print("\nAfter dropout:")
print(layer_dropout)
```

**Interpretation:**

Dropout randomly disables some neurons during training.

This prevents the network from depending too strongly on specific neurons and can reduce overfitting.

## 2. Batching

Instead of updating the network after every single observation, training data can be processed in groups called batches.

Mini-batch training balances training speed and update frequency.

```python
data = np.arange(20)

batch_size = 5

for start in range(
    0,
    len(data),
    batch_size
):
    batch = data[
        start:start + batch_size
    ]

    print("Batch:", batch)
```
