# Chapter 13 - Introducing Automatic Optimization

This chapter introduces the idea of building a small deep learning framework.

The goal is to automate repeated operations such as layer computation, parameter updates, and optimization.

```python
import numpy as np

class Linear:

    def __init__(
        self,
        input_size,
        output_size
    ):

        self.weights = (
            np.random.randn(
                input_size,
                output_size
            ) * 0.1
        )

        self.input = None


    def forward(self, input_data):

        self.input = input_data

        return input_data.dot(
            self.weights
        )


layer = Linear(
    3,
    2
)

x = np.array([
    [1.0, 2.0, 3.0]
])

output = layer.forward(x)

print(output)
```

## Optimizer

An optimizer updates model parameters using calculated gradients.

Separating layers and optimizers makes neural network code easier to organize and reuse.

```python
class SGD:

    def __init__(
        self,
        learning_rate=0.01
    ):

        self.learning_rate = (
            learning_rate
        )


    def update(
        self,
        parameter,
        gradient
    ):

        parameter -= (
            self.learning_rate
            * gradient
        )

        return parameter


optimizer = SGD(
    learning_rate=0.1
)

weight = np.array([1.0])
gradient = np.array([0.5])

weight = optimizer.update(
    weight,
    gradient
)

print(weight)
```
