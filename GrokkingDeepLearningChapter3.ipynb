# Chapter 3 - Introduction to Neural Prediction: Forward Propagation

This chapter introduces neural prediction and forward propagation.

A neural network combines input values and weights to produce a prediction.

```python
def neural_network(input_value, weight):
    prediction = input_value * weight
    return prediction

weight = 0.1
input_value = 8.5

prediction = neural_network(input_value, weight)

print("Prediction:", prediction)
```

**Interpretation:**

The neural network multiplies the input by the weight.

The weight determines how strongly the input contributes to the prediction.

## Multiple Inputs

A neural network can receive several input values.

Each input has a corresponding weight. The weighted inputs are combined to create the final prediction.

```python
def weighted_sum(a, b):
    output = 0

    for i in range(len(a)):
        output += a[i] * b[i]

    return output


def neural_network(inputs, weights):
    return weighted_sum(inputs, weights)


inputs = [8.5, 0.65, 1.2]
weights = [0.1, 0.2, 0.0]

prediction = neural_network(inputs, weights)

print("Prediction:", prediction)
```

```python
import numpy as np

inputs = np.array([8.5, 0.65, 1.2])
weights = np.array([0.1, 0.2, 0.0])

prediction = np.dot(inputs, weights)

print("Prediction:", prediction)
```

When several inputs are used, the dot product provides an efficient way to calculate the weighted sum.
