# Chapter 5 - Learning Multiple Weights at a Time

This chapter extends gradient descent to neural networks with multiple inputs and multiple weights.


```python
import numpy as np

inputs = np.array([1.0, 0.5, -1.0])
weights = np.array([0.2, 0.3, -0.1])

goal_prediction = 1.0
alpha = 0.1

for iteration in range(20):

    prediction = np.dot(inputs, weights)

    error = (
        prediction - goal_prediction
    ) ** 2

    delta = (
        prediction - goal_prediction
    )

    weight_deltas = (
        inputs * delta
    )

    weights -= (
        alpha * weight_deltas
    )

    print(
        iteration,
        "Prediction:",
        round(prediction, 4),
        "Error:",
        round(error, 4)
    )
```

**Interpretation:**

Each input has its own weight.

During training, all weights can be updated at the same time based on their contribution to the prediction error.

## Gradient Descent with Multiple Outputs

A neural network can also produce several predictions at the same time.

Each output can have its own error and weight adjustments.

```python
inputs = np.array([1.0, 0.5, -1.0])

weights = np.array([
    [0.1, 0.2, 0.1],
    [0.3, 0.1, -0.2],
    [-0.1, 0.2, 0.4]
])

predictions = np.dot(
    inputs,
    weights
)

print(predictions)
```
