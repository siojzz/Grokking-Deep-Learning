# Chapter 2 - Fundamental Concepts: How Do Machines Learn?

This chapter introduces the relationship between artificial intelligence, machine learning, and deep learning.

It also explains supervised learning, unsupervised learning, parametric learning, and nonparametric learning.

## 1. Artificial Intelligence, Machine Learning, and Deep Learning

Artificial Intelligence is the broader idea of creating machines that perform tasks associated with intelligence.

Machine Learning is a part of AI where machines learn patterns from data.

Deep Learning is a part of machine learning that mainly uses artificial neural networks.

## 2. Supervised Learning

Supervised learning uses data where the correct output is already known.

The model learns a relationship between the input and the expected output.

Examples include predicting house prices and classifying emails as spam or not spam.

## 3. Unsupervised Learning

Unsupervised learning works with data that does not have known target labels.

The goal is usually to discover patterns, structures, or groups inside the data.

## 4. Parametric and Nonparametric Learning

A parametric model learns a fixed set of parameters from data.

A neural network is a parametric model because it learns parameters called weights.

Nonparametric methods can use the training data more directly without representing the solution using only a fixed number of parameters.

```python
input_value = 2
weight = 0.5

prediction = input_value * weight

print("Prediction:", prediction)
```

**Interpretation:**

This simple example represents parametric learning.

The weight is a parameter that controls how the input affects the prediction.
