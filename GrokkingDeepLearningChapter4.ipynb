# Chapter 4 - Introduction to Neural Learning: Gradient Descent

This chapter explains how a neural network learns.

Learning happens by measuring prediction error and adjusting the weights to reduce that error.

```python
weight = 0.5
input_value = 0.5
goal_prediction = 0.8

prediction = input_value * weight

error = (prediction - goal_prediction) ** 2

print("Prediction:", prediction)
print("Error:", error)
```

## Gradient Descent

Gradient descent adjusts a weight in the direction that reduces the prediction error.

The weight delta tells us how much the weight should change.

```python
weight = 0.5
input_value = 0.5
goal_prediction = 0.8
alpha = 0.1

for iteration in range(20):

    prediction = input_value * weight

    error = (prediction - goal_prediction) ** 2

    direction_and_amount = (
        prediction - goal_prediction
    ) * input_value

    weight = weight - (
        alpha * direction_and_amount
    )

    print(
        iteration,
        "Prediction:",
        round(prediction, 4),
        "Error:",
        round(error, 4)
    )
```

**Interpretation:**

The prediction gradually moves closer to the target value.

At the same time, the error becomes smaller.

This shows how gradient descent allows a neural network to learn by changing its weights.
