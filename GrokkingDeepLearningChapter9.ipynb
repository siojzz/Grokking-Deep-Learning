# Chapter 9 - Modeling Probabilities and Nonlinearities

This chapter introduces activation functions and explains why nonlinear functions are important in deep neural networks.

```python
import numpy as np

def sigmoid(x):
    return 1 / (
        1 + np.exp(-x)
    )

values = np.array([
    -3, -1, 0, 1, 3
])

print(
    sigmoid(values)
)
```

```python
def tanh(x):
    return np.tanh(x)

print(
    tanh(values)
)
```

## Softmax

Softmax converts several numerical scores into probabilities.

The probabilities sum to 1 and can be used for classification with multiple classes.

```python
def softmax(x):

    exp = np.exp(
        x - np.max(x)
    )

    return (
        exp / exp.sum()
    )


scores = np.array([
    2.0,
    1.0,
    0.1
])

probabilities = softmax(
    scores
)

print(probabilities)
print(
    "Sum:",
    probabilities.sum()
)
```

Functions such as sigmoid and tanh allow networks to represent more complex relationships.
